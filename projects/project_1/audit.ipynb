```python
import csv 
import json

def validate_record(row):
    errors = []
    if not row['id'].isdigit():
        errors.append('invalid_id')
    if not row['age']:
        errors.append('missing_age')
    elif not row['age'].isdigit():
        errors.append('invalid_age')
    if not row['department']:
        errors.append('missing_department')
    return errors    

total = 0
valid_records = 0
invalid_records = 0

all_errors = []
department_count = {}
errors_count = {}

with open('data.csv',mode='r') as file:
    reader = csv.DictReader(file)

    for row in reader:
        total += 1
        errors = validate_record(row)

        if errors:
            for error in errors:
                errors_count[error] = errors_count.get(error, 0) + 1
            invalid_records += 1
            all_errors.append({
                'id':row["id"],
                'errors':errors
            })
        else:
            valid_records += 1
            department = row['department']
            department_count[department] = department_count.get(department, 0) + 1


print("\nAudit Summary")
print(f"Total record:{total}")
print(f"Valid records:{valid_records}")
print(f"Invalid records:{invalid_records}")

print("\nInvalid records")
for error in all_errors:
    print(error)
print("\nDepartment summary")
for department,count in department_count.items():
    print(f"{department}:{count}")    

print("\nValidation Error Summary")

for error, count in errors_count.items():
    print(f"{error}: {count}")    

summary = {
    'total_records': total,
    'valid_records': valid_records,
    'invalid_records': invalid_records,
    'invalid_records_detail': all_errors,
    'department_summary': department_count,
    'error_summary': errors_count
}

with open('summary.json', mode='w') as file:
    json.dump(summary, file, indent=4)
```

```python
import csv
import json


def read_records(filename):
    with open(filename, mode='r') as file:
        reader = csv.DictReader(file)

        for row in reader:
            yield row


def validate_record(row):
    errors = []

    if not row['id'].isdigit():
        errors.append('invalid_id')

    if not row['age']:
        errors.append('missing_age')
    elif not row['age'].isdigit():
        errors.append('invalid_age')

    if not row['department']:
        errors.append('missing_department')

    return errors


total = 0
valid_records = 0
invalid_records = 0

all_errors = []
department_count = {}
errors_count = {}


# Read one record at a time
for row in read_records('data.csv'):

    total += 1

    errors = validate_record(row)

    if errors:

        invalid_records += 1

        # Count each type of error
        for error in errors:
            errors_count[error] = errors_count.get(error, 0) + 1

        # Store invalid record details
        all_errors.append({
            'id': row['id'],
            'errors': errors
        })

    else:

        valid_records += 1

        # Group valid records by department
        department = row['department']

        department_count[department] = (
            department_count.get(department, 0) + 1
        )


# Terminal output

print("\nAudit Summary")

print(f"Total records: {total}")
print(f"Valid records: {valid_records}")
print(f"Invalid records: {invalid_records}")


print("\nInvalid Records")

for error in all_errors:
    print(error)


print("\nDepartment Summary")

for department, count in department_count.items():
    print(f"{department}: {count}")


print("\nValidation Error Summary")

for error, count in errors_count.items():
    print(f"{error}: {count}")


# Create JSON summary

summary = {
    'total_records': total,
    'valid_records': valid_records,
    'invalid_records': invalid_records,
    'invalid_records_detail': all_errors,
    'department_summary': department_count,
    'error_summary': errors_count
}


with open('summary.json', mode='w') as file:
    json.dump(summary, file, indent=4)

print("\nJSON report created: summary.json")
```
