```python
def validate_records(record):
    valid = True
    if not record['id'].isdigit():
        print(f"Invalid ID: {record['id']}")
        valid = False

    if not record['age'].isdigit():
        print(f"Invalid Age: {record['age']} -- ID: {record['id']}")
        valid = False

    if not record['department']:
        print(f"Missing Department for ID: {record['id']}")
        valid = False
    return valid

total = 0
valid_records = 0
invalid_records = 0

with open('data.csv', mode='r') as file:
    reader = csv.DictReader(file)

    for row in reader:
        total += 1
        if validate_records(row):
            valid_records += 1
        else:
            invalid_records += 1

print()
print(f"Total records: {total}")
print(f"Valid records: {valid_records}")
print(f"Invalid records: {invalid_records}")
           
```

```python
import csv


def validate_record(row):
    errors = []

    if not row['id'].isdigit():
        errors.append('invalid_id')

    if not row['age']:
        errors.append('missing_age')
    elif not row['age'].isdigit():
        errors.append('invalid_age')

    if not row['department']:
        errors.append('missing_department')

    return errors


total = 0
valid_records = 0
invalid_records = 0

all_errors = []


with open('data.csv', mode='r') as file:
    reader = csv.DictReader(file)

    for row in reader:
        total += 1

        errors = validate_record(row)

        if errors:
            invalid_records += 1

            # Print errors for this specific row
            print(f"ID {row['id']} is invalid: {errors}")

            # Add this row's errors to the main list
            all_errors.append({
                'id': row['id'],
                'errors': errors
            })

        else:
            valid_records += 1


print()
print(f"Total records: {total}")
print(f"Valid records: {valid_records}")
print(f"Invalid records: {invalid_records}")

print()
print("All Errors:")

for error in all_errors:
    print(error)
```
