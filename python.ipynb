# Python Memory Management

Chapter 1 — What Does "Memory Management" Mean?

Before learning reference counting or garbage collection, understand the actual problem.

Every program needs memory to store things like:




```python
name = "Vishal"
age = 21
numbers = [1, 2, 3]
```

Python creates objects to represent these values.

Conceptually:

name ─────► "Vishal"

age ──────► 21

numbers ──► [1, 2, 3]

The Python runtime has to answer two basic questions:

Where should objects be stored?
When can their memory be reclaimed?

Memory management is about handling the lifecycle of these objects.

Chapter 2 — Variables Do Not Own Memory in the Way Beginners Think

Consider:

x = 10

A beginner might imagine:

x

┌─────────┐
│   10    │
└─────────┘


This is a useful beginner abstraction, but it is not the best mental model for Python.

A better model is:

x ─────────► Integer Object: 10

Now:

y = x

becomes:


x ───────┐
         ▼
        10
         ▲
         │
y ───────┘


There is one object and two names.

This is why understanding references is necessary before understanding memory management.

Chapter 3 — Object Lifetime

Every Python object has a lifetime.

Conceptually:

Object Created

      ↓

Object Used

      ↓

Object Referenced

      ↓

References Removed

      ↓

Object Becomes Unreachable

      ↓
      
Memory Can Be Reclaimed

```python
x = [1, 2, 3]
```

An object is created.

Then:

```python
y = x
```

Now two names refer to it.

Then:

```python
del x
```

The list still exists because:

y ─────► [1, 2, 3]

Now:

```python
del y
```

If no other references exist, the object becomes unreachable.

Conceptually:

No references

     ↓

Object is unreachable

     ↓
     
Python can reclaim it

This leads us to one of the most important concepts in CPython:

Reference Counting

Chapter 4 — Reference Counting

Conceptually:

Reference count → 0

        ↓

Object becomes unreachable

        ↓
        
CPython can deallocate it

This is the basic idea behind reference counting.

Chapter 5 — del Does Not Mean "Delete the Object"

del x removes the name binding.

It does not necessarily destroy the object.

This distinction is critical:

del variable

       ↓

Remove a reference/name

       ↓
       
Not necessarily destroy object

Chapter 6 — del and Object Lifetime

```python
x = [1, 2, 3]
del x
```

if x was the only reference to the object, the object may become unreachable.

But:

```python
x = [1, 2, 3]
y = x

del x
```

The object remains reachable through y.

So:

del x

means:

Remove the name x from its namespace.

It does not mean:

Forcefully destroy the object

Chapter 7 — Measuring Reference Counts

Python provides:


```python
import sys
```

You can inspect reference counts with:

```python
sys.getrefcount()
```

Example:

```python
import sys

x = []

print(sys.getrefcount(x))
```

But don't immediately conclude:

"The object has exactly two references."

There is an important reason.

When you call:

```python
sys.getrefcount(x)
```

the function itself temporarily receives a reference to the object.

So the reported number is typically one higher than the count you might expect from your own references.

Example:

```python
x = []
y = x
```

You may see a count reflecting:

x
y
temporary function argument

This makes sys.getrefcount() useful for experimentation, but not something you should casually rely on for application logic.

Chapter 8 — Reference Counting Is Not Enough

```python
a = []
b = []

a.append(b)
b.append(a)
```

The objects reference each other.

Now:

```python
del a
del b
```

From the program's namespace, the names are gone.

But conceptually:

The objects still reference each other.

Their reference counts may not reach zero.

This creates a reference cycle.

Reference counting alone cannot solve this.

This is why Python also has:

Garbage Collection

Chapter 9 — Garbage Collection

Python's garbage collector can detect certain groups of objects that are no longer reachable from the program but still reference one another.

Python provides the gc module:

```python
import gc
```

You can inspect whether automatic garbage collection is enabled:

```python
print(gc.isenabled())
```

You can manually trigger collection:

```python
gc.collect()
```

But in normal application code, you generally should not manually call gc.collect() everywhere.

The garbage collector is designed to operate automatically.

Chapter 10 — Reference Counting vs Garbage Collection

Reference Counting. ---- Tracks references to objects

Cyclic Garbage Collector --- Handles unreachable reference cycles

Chapter 11 — Reference Cycles

Let's create a simple cycle:

```python
a = []
a.append(a)
```

The list contains a reference to itself.

Now:


```python
del a
```

The name a disappears.

But the object has a self-reference.

This is a cycle.

The cyclic garbage collector can detect such unreachable cyclic structures.

This is why Python's memory management is more complicated than:

"Reference count reaches zero, object disappears."

That statement is incomplete.

Chapter 12 — Generational Garbage Collection

Python's cyclic garbage collector uses a generational approach.

The underlying idea is based on an observation:

Objects that survive for a long time are often more likely to continue surviving.

So objects are conceptually organized by age/generation.

The garbage collector focuses more frequently on newer objects and less frequently on older ones.
