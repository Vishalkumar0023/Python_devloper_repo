2. Mutable VS Immutable Objects

--Immutable Objects -

int ,float ,str ,complex ,bool ,tuple , bytes ,frozenset ,None

Mutable Types-

list
dict	
set	
bytearray	
most class instances

```python
x = 10

print(id(x))

x = 20

print(id(x))
```

```python
numbers = [10,20]

print(id(numbers))

numbers.append(30)

print(id(numbers))
```

```python
# string are immutable 


text = "Python"

text[0] = "J"
```

Python refuses to modify them.


Correct approach

```python
text = "Python"

text = "J" + text[1:]

print(text)
print("A completely new string object is created.")
```

```python
student = {"name":"Alex"}

print(id(student))

student["age"] = 20

print(id(student))
```
