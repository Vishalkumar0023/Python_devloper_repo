== Can Be Customized


This is an advanced but important point.

When you write:

a == b

Python generally invokes the object's equality behavior.

For user-defined classes, you can define:



```python
print("__eq__()")
```

Example:

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __eq__(self, other):
        return self.name == other.name
```

```python
student1 = Student("Vishal")
student2 = Student("Vishal")
```

```python
print(student1 == student2)
```

because their equality behavior says they are equal.

```python
print(student1 is student2)
```

because they are separate objects.

== is about logical equality.

is is about physical/object identity.
